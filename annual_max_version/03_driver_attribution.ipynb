```python
def select_utci_max(ds):
    """Return the values at the time of UTCI max."""
    # Find index of UTCI max for this year
    idx = ds['UTCI'].argmax(dim='time')

    # Select values at that time across all variables
    return ds.isel(time=idx)
```

```python
# Process each year individually to save memory
mask = ds['UTCI'].isel(time=0).notnull()
print("Processing years individually...")

# Get unique years
years = pd.date_range(start='1970', end='2022', freq='Y').year

# Initialize lists to store annual max data
annual_max_data = []

for year in years:
    print(f"Processing year {year}...")
    
    # Select data for this year
    year_data = ds.sel(time=slice(str(year), str(year)))
    
    # Skip if no data for this year
    if year_data.time.size == 0:
        print(f"No data for year {year}, skipping...")
        continue
    
    # Fill NaN values 
    year_data_filled = year_data.fillna(-9999)
    
    # Compute the annual max for this year
    year_max = select_utci_max(year_data_filled)
    
    # Apply mask and add year coordinate
    year_max = year_max.where(mask).expand_dims('year')
    year_max = year_max.assign_coords(year=[year])
    
    # Store the result
    annual_max_data.append(year_max)
    
    # Clear memory
    del year_data, year_data_filled, year_max

# Combine all years
print("Combining all years...")
ds_annual_max = xr.concat(annual_max_data, dim='year')

# Clear the list to free memory
del annual_max_data

print("Processing complete!")
```
